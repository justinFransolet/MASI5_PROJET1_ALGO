### 4. Données manquantes

Visualiser les données manquantes et les remplacer.

```python
msno.matrix(housing_data)
plt.show()

# Nombre de valeurs manquantes avant imputation
missing_before = housing_data.isna().sum()

# Identification des colonnes
numeric_features = housing_data.select_dtypes(include=["number"]).columns
categorical_features = housing_data.select_dtypes(exclude=["number"]).columns

# Prétraitement des colonnes numériques
numeric_transformer = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=2))
])

# Prétraitement des colonnes catégorielles
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("categorical", categorical_transformer, categorical_features)
    ],
    verbose_feature_names_out=False,
)

housing_processed = preprocessor.fit_transform(housing_data)

# Récupération des noms de colonnes après OneHotEncoding
feature_names = preprocessor.get_feature_names_out()
housing_processed_df = pd.DataFrame(
    housing_processed, columns=feature_names, index=housing_data.index
)

# Nombre de valeurs manquantes après imputation
missing_after = housing_processed_df.isna().sum()

# Tableau comparatif
comparison = pd.DataFrame({
    "Avant": missing_before,
    "Après": missing_after,
    "Valeurs remplacées": missing_before - missing_after
})

print("\nComparaison avant/après :")
print(comparison[comparison["Avant"] > 0])
```

Il ne manquait que 207 valeurs dans la colonne `total_bedrooms`.
Nous les comblons avec la méthode **KNN**.
Car le faire avec une méthode moyenne ou médiane va créer un pic artificiel.

## Modèle

```python
# Séparation des caractéristiques et de la cible
X = housing_processed_df.drop(columns=["median_house_value"])
y = housing_processed_df["median_house_value"]

# Identification des colonnes
numeric_features = X.select_dtypes(include=["number"]).columns
categorical_features = X.select_dtypes(exclude=["number"]).columns

# Prétraitement des colonnes catégorielles
categorical_transformer = Pipeline(steps=[
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("categorical", categorical_transformer, categorical_features)
    ]
)

# Modèle complet : prétraitement puis arbre de décision
model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", DecisionTreeRegressor(
        max_depth=18,
        random_state=50
    ))
])

# Séparation apprentissage/test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=50
)

# Entraînement
model.fit(X_train, y_train)

# Prédictions
y_pred = model.predict(X_test)

# Évaluation
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"RMSE : {rmse:.2f}")
print(f"Score R² : {r2:.4f}")
```
