### 4. Données manquantes

Visualiser les données manquantes et les remplacer.

```python
# Copie avant traitement
housing_before = housing_data.copy()

msno.matrix(housing_data)
plt.show()

# Nombre de valeurs manquantes avant imputation
missing_before = housing_before.isna().sum()

# Colonnes numériques remplacer par la médiane
numerical_columns = housing_data.select_dtypes(include="number").columns
medians = housing_data[numerical_columns].median()
housing_data[numerical_columns] = housing_data[numerical_columns].fillna(medians)

# Colonnes catégorielles remplacer par l'occurence la plus présente
categorical_columns = housing_data.select_dtypes(
    include=["object", "string", "category"]
).columns

for column in categorical_columns:
    housing_data[column] = housing_data[column].fillna(
        housing_data[column].mode().iloc[0]
    )

# Nombre de valeurs manquantes après imputation
missing_after = housing_data.isna().sum()

# Tableau comparatif
comparison = pd.DataFrame({
    "Avant": missing_before,
    "Après": missing_after,
    "Valeurs remplacées": missing_before - missing_after
})

print("\nComparaison avant/après :")
print(comparison[comparison["Avant"] > 0])
```

## Test

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Séparation des caractéristiques et de la cible
X = housing_data.drop(columns=["median_house_value"])
y = housing_data["median_house_value"]

# Identification des colonnes
numeric_features = X.select_dtypes(include=["number"]).columns
categorical_features = X.select_dtypes(exclude=["number"]).columns

# Prétraitement des colonnes numériques
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

# Prétraitement des colonnes catégorielles
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("categorical", categorical_transformer, categorical_features)
    ]
)

# Modèle complet : prétraitement puis arbre de décision
model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", DecisionTreeRegressor(
        max_depth=11,
        random_state=25
    ))
])

# Séparation apprentissage/test
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=25
)

# Entraînement
model.fit(X_train, y_train)

# Prédictions
y_pred = model.predict(X_test)

# Évaluation
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"RMSE : {rmse:.2f}")
print(f"Score R² : {r2:.4f}")
```
