```python
msno.matrix(housing_data)
plt.show()

missing_before = X_train.isna().sum()

# Identification des colonnes
numeric_features = X.select_dtypes(include=["number"]).columns
categorical_features = X.select_dtypes(exclude=["number"]).columns

# Prétraitement des colonnes numériques
numeric_transformer = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=5))
])

# Prétraitement des colonnes catégorielles
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_transformer, numeric_features),
        ("categorical", categorical_transformer, categorical_features)
    ],
    verbose_feature_names_out=False,
)

X_train_processed = preprocessor.fit_transform(X_train)

# Récupération des noms de colonnes après OneHotEncoding
feature_names = preprocessor.get_feature_names_out()
X_train_processed_df = pd.DataFrame(
    X_train_processed, columns=feature_names, index=X_train.index
)
X_train_processed_df.columns = [
    str(col).replace("<", "lt_").replace("[", "_").replace("]", "_")
    for col in X_train_processed_df.columns
]

# Afficher les nombres de données imputées
print(f"Valeurs manquantes avant imputation: {housing_data.isna().sum()}")
print(f"Valeurs manquantes après imputation: {X_train.isna().sum()}")
```

# Modèles et hyperparamètres

Les modèles comparés sont `DecisionTreeRegressor`, `RandomForestRegressor`, `XGBRegressor` et `LGBMRegressor`.

Les hyperparamètres explorés contrôlent principalement la complexité des arbres, la régularisation, le nombre d'arbres et la diversité des arbres. Les valeurs optimales seront déterminées par Optuna à partir de la RMSE en validation croisée.

**Attention :** toutes les étapes de prétraitement sont incluses dans le pipeline évalué par validation croisée. Ainsi, chaque fold ajuste l'imputation et l'encodage uniquement sur sa partie d'entraînement.

```python
def create_model(trial, model_name):
    if model_name == "DecisionTree":
        return DecisionTreeRegressor(
            max_depth=trial.suggest_int("max_depth", 3, 25),
            min_samples_split=trial.suggest_int("min_samples_split", 2, 30),
            min_samples_leaf=trial.suggest_int("min_samples_leaf", 1, 20),
            max_features=trial.suggest_categorical("max_features", [None, "sqrt", "log2"]),
            random_state=42
        )

    elif model_name == "RandomForest":
        return RandomForestRegressor(
            n_estimators=trial.suggest_int("n_estimators", 100, 500),
            max_depth=trial.suggest_int("max_depth", 5, 30),
            min_samples_split=trial.suggest_int("min_samples_split", 2, 20),
            min_samples_leaf=trial.suggest_int("min_samples_leaf", 1, 10),
            max_features=trial.suggest_categorical("max_features", ["sqrt", "log2", 0.7, 1.0]),
            random_state=42,
            n_jobs=-1
        )

    elif model_name == "XGBoost":
        return XGBRegressor(
            n_estimators=trial.suggest_int("n_estimators", 200, 1000),
            learning_rate=trial.suggest_float("learning_rate", 0.01, 0.2, log=True),
            max_depth=trial.suggest_int("max_depth", 3, 10),
            subsample=trial.suggest_float("subsample", 0.6, 1.0),
            colsample_bytree=trial.suggest_float("colsample_bytree", 0.6, 1.0),
            reg_alpha=trial.suggest_float("reg_alpha", 1e-3, 10.0, log=True),
            reg_lambda=trial.suggest_float("reg_lambda", 1e-3, 10.0, log=True),
            random_state=42,
            n_jobs=-1
        )

    elif model_name == "LightGBM":
        return LGBMRegressor(
            n_estimators=trial.suggest_int("n_estimators", 200, 1000),
            learning_rate=trial.suggest_float("learning_rate", 0.01, 0.2, log=True),
            num_leaves=trial.suggest_int("num_leaves", 20, 150),
            max_depth=trial.suggest_int("max_depth", 3, 12),
            min_child_samples=trial.suggest_int("min_child_samples", 10, 100),
            subsample=trial.suggest_float("subsample", 0.6, 1.0),
            colsample_bytree=trial.suggest_float("colsample_bytree", 0.6, 1.0),
            random_state=42,
            n_jobs=-1,
            verbose=-1
        )

    raise ValueError(f"Modèle inconnu : {model_name}")
```

## Entraînement et optimisation

Nous utilisons Optuna pour rechercher les hyperparamètres. La fonction objectif évalue **le pipeline complet** par validation croisée : le prétraitement est donc réappris à l'intérieur de chaque fold, sans utiliser les observations de validation du fold pour ajuster l'imputation ou l'encodage.

Le jeu de test est conservé à l'écart jusqu'à la comparaison finale.

```python
# Nombre de test autorisé par modèle
models_trials = {
    "DecisionTree": 30,
    "RandomForest": 25,
    "XGBoost": 40,
    "LightGBM": 40
}

studies = {}
best_pipelines = {}
results = []

for model_name, n_trials in models_trials.items():
    print(f"Optimisation de {model_name} ({n_trials} essais)...")

    def objective(trial):
        model = create_model(trial, model_name)
        scores = cross_val_score(
            model, X_train_processed_df, y_train,
            cv=5, scoring="neg_root_mean_squared_error", n_jobs= 1
        )
        return -scores.mean()

    # Étude des hyperparamètres
    study = optuna.create_study(
        direction="minimize",
        study_name=model_name,
        sampler=optuna.samplers.TPESampler(seed=42)
    )
    study.optimize(objective, n_trials=n_trials)
    studies[model_name] = study

    # Refaire un modèle avec les meileurs hyperparamètres
    best_model = create_model(optuna.trial.FixedTrial(study.best_params), model_name)

    full_pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("regressor", best_model)
    ])

    # Entraînement final
    full_pipeline.fit(X_train, y_train)

    best_pipelines[model_name] = full_pipeline
    results.append({
        "Modèle": model_name,
        "CV RMSE": float(study.best_value),
        "Meilleurs Paramètres": study.best_params
    })

print("\nOptimisation terminée !")
```

# Évaluation sur les jeux d'entraînement et de test

Expliquer la CV RMSE.

La RMSE mesure l'erreur de prédiction dans l'unité de la cible (dollars). Le R² mesure la part de variance expliquée par le modèle. Un écart important entre les performances d'entraînement et de test peut signaler du surapprentissage, mais il n'existe pas de seuil universel : il faut interpréter cet écart avec les scores de validation croisée et le contexte du problème.

```python
df_results = pd.DataFrame(results).sort_values( by="CV RMSE", ascending=True).reset_index(drop=True)

best_model_name = df_results.loc[0, "Modèle"]
winner_pipeline = best_pipelines[best_model_name]

print("Modèle retenu sur la base de la validation croisée :", best_model_name)
print("Meilleurs paramètres :", df_results.loc[0, "Meilleurs Paramètres"])

# Évaluation finale de chaque modèle sur base du dataset de test.
evaluation_rows = []
for model_name, pipeline in best_pipelines.items():
    y_pred_train = pipeline.predict(X_train)
    y_pred_test = pipeline.predict(X_test)

    rmse_train = np.sqrt(mean_squared_error(y_train, y_pred_train))
    r2_train = r2_score(y_train, y_pred_train)
    rmse_test = np.sqrt(mean_squared_error(y_test, y_pred_test))
    r2_test = r2_score(y_test, y_pred_test)

    evaluation_rows.append({
        "Modèle": model_name,
        "RMSE entraînement": rmse_train,
        "R2 entraînement": r2_train,
        "RMSE test": rmse_test,
        "R2 test": r2_test,
        "Écart R2 (train - test)": r2_train - r2_test
    })

    print(f"\n=== {model_name} ===")
    print(f"Entraînement : RMSE = {rmse_train:,.2f} $ | R² = {r2_train:.4f}")
    print(f"Test         : RMSE = {rmse_test:,.2f} $ | R² = {r2_test:.4f}")
    print(f"Écart de R²  : {r2_train - r2_test:.4f}")

df_evaluation = pd.DataFrame(evaluation_rows).sort_values(
    "RMSE test"
).reset_index(drop=True)

display(df_results)
display(df_evaluation)
```

## Interprétation de l'évaluation

Le modèle retenu est celui qui minimise la RMSE moyenne de validation croisée sur le jeu d'entraînement. Les métriques calculées sur le jeu de test donnent ensuite une estimation de sa capacité de généralisation sur des données non utilisées pendant l'ajustement.

Ne pas conclure à une bonne généralisation à partir d'un seuil arbitraire d'écart de R². Examiner conjointement les RMSE, les R², leur écart entre entraînement et test, ainsi que la variabilité des scores entre les folds.
